# Vocal separation with Spleeter and Demucs

Separate a local music file into vocals and accompaniment using either pretrained tool. Put a file named `song.mp3` in this directory or change `audio_path` below. Install Spleeter and Demucs in compatible environments before running their respective cells; the first run may download model weights. Outputs are written under `output/`.

This is a small command-line example. The larger comparison notebook is `main_2.ipynb`. These cells have not been rerun after cleanup.

**Data:** No fixed dataset. Supply a local recording as `song.mp3`; the evaluation cells also need separate reference stems when calculating separation metrics.

```python
from pathlib import Path
import subprocess
import sys
from IPython.display import Audio, display

audio_path = Path("song.mp3")
if not audio_path.is_file():
    raise FileNotFoundError(f"Add the input track: {audio_path}")
```

## Spleeter

```python
spleeter_dir = Path("output/spleeter_output")
subprocess.run([sys.executable, "-m", "spleeter", "separate", "-p", "spleeter:2stems",
                "-o", str(spleeter_dir), str(audio_path)], check=True)
spleeter_stems = spleeter_dir / audio_path.stem
for name in ("vocals", "accompaniment"):
    path = spleeter_stems / f"{name}.wav"
    print(name, path)
    display(Audio(filename=str(path)))
```

## Demucs

```python
demucs_dir = Path("output/demucs_output")
subprocess.run([sys.executable, "-m", "demucs", "--two-stems=vocals",
                "-o", str(demucs_dir), str(audio_path)], check=True)
demucs_stems = demucs_dir / "htdemucs" / audio_path.stem
for name in ("vocals", "no_vocals"):
    path = demucs_stems / f"{name}.wav"
    print(name, path)
    display(Audio(filename=str(path)))
```
