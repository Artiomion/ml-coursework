# Lesson 12: Graph theory

A graph consists of vertices and edges. Its structure describes which vertices are connected. Positions in a drawing need not represent coordinates in the data.

## Incidence and adjacency

An edge is incident to each endpoint. Adjacent vertices share an edge. Adjacent edges share a vertex.

![Illustration of incidence and adjacency](pic/196e989ffc3dad638f48799367d8a38e.jpg)

## Loops

A loop connects a vertex to itself. In an undirected graph it contributes two to that vertex’s degree.

![Illustration of loops](pic/b6cd3747619e537188ab6e7399e45328.png)

## Parallel edges

Parallel edges share both endpoints. A multigraph permits parallel edges. Terminology for graphs that also permit loops varies.

![Illustration of parallel edges](pic/b8fdd3cf189bc899b66d6096c9cb10ee.jpg)

## Degree

The degree of a vertex counts its incident edge ends. An isolated vertex has degree zero; a leaf has degree one.

![Illustration of degree](pic/d7ee08a28563fe6710a2c414926c2f28.jpg)

## Subgraphs

A subgraph consists of selected vertices and edges from another graph.

![Illustration of subgraphs](pic/4682ae118793c500ee384f37d644aec5.png)

## Complete graphs

In a complete graph, every pair of distinct vertices is connected by an edge.

![Illustration of complete graphs](pic/9abc15721c29dbc8a9dc61eaae355be3.png)

## Regular graphs

Every vertex of a regular graph has the same degree.

![Illustration of regular graphs](pic/594284c2e545f1f07807b85810842480.png)

## Bipartite graphs

The vertices can be divided into two sets such that every edge joins vertices in different sets.

![Illustration of bipartite graphs](pic/7e2517bdc2b557d514a5618661f708f5.png)

## Planar graphs

A graph is planar if it can be drawn in the plane with no crossing edges.

![Illustration of planar graphs](pic/2e270ead9df3601b882cb088638bc2f5.png)

## Walks, trails, and paths

A walk follows adjacent edges. A trail does not repeat edges; a simple path does not repeat vertices. Length counts edges.

![Illustration of walks, trails, and paths](pic/6f6ed7afe8bee69813675589384164e4.jpg)

## Cycles

A cycle returns to its starting vertex. Conventions about loops and repeated vertices depend on the type of graph.

![Illustration of cycles](pic/08147a9b12c69345e0dab83efdc21c9c.png)

## Eulerian circuits

An Eulerian circuit traverses each edge exactly once. A connected undirected graph has one when every vertex has even degree.

![Illustration of eulerian circuits](pic/8a5306eccdde68a02a324c894a1281c7.jpg)

## Hamiltonian cycles

A Hamiltonian cycle visits each vertex exactly once before returning to the start.

![Illustration of hamiltonian cycles](pic/97ef417f30dce2a799e8bc3105a41270.jpg)

## Weighted graphs

Weights can represent costs or distances. Algorithms can use them to find a minimum-cost path.

![Illustration of weighted graphs](pic/eed345683a3e8b814ac6415230cbc9e7.jpg)

## Trees

A connected undirected graph has a path between every pair of vertices. A tree is connected and acyclic.

![Illustration of trees](pic/2b91a68bb72d006b9cd9ae9fbe902f62.jpg)

## Forests

A forest is an acyclic undirected graph.

![Illustration of forests](pic/50132d3ba8ae6672f00b96d2b797eac7.jpg)

## Directed graphs

A directed edge, or arc, has a direction.

![Illustration of directed graphs](pic/12777d85bf1d6e441663e9f380ee0607.jpg)

## Degrees in directed graphs

Indegree counts incoming arcs and outdegree counts outgoing arcs. A source has indegree zero; a sink has outdegree zero.

![Illustration of degrees in directed graphs](pic/d25560cadcdd2aa19c01f7665b0c2647.png)

## Connected components

A connected component is a maximal connected set of vertices.

![Illustration of connected components](pic/801d7e0a056fd759f7d097775d936029.jpg)

## Directed connectivity

Strong connectivity requires paths in both directions. Weak connectivity ignores arc direction.

![Illustration of directed connectivity](pic/67225fd178e6a80be1320eea86471061.jpg)

## Bridges

A bridge is an edge whose removal increases the number of connected components.

![Illustration of bridges](pic/21aedb8f6c9e0b8d7d4d36ff4a25d309.jpg)
