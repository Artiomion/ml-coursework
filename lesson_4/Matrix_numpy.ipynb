Run the cells in order with NumPy installed. Start by creating a 3 × 3 array from nested lists.

**Data:** No external dataset. Matrices are defined in the notebook.

```python
import numpy as np

matrix_A = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])
```

## Special matrices

The shape tells NumPy how many rows and columns to create. An identity matrix has ones on its diagonal and zeros elsewhere.

```python
zeros = np.zeros((3, 3))
ones = np.ones((2, 4))
eye = np.eye(3)
random = np.random.rand(3, 3)
```

## Elementwise arithmetic

For these two arrays of the same shape, each result comes from applying the operator to entries in the same position.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])


C = A + B


D = A - B


E = A * B


F = A / B
```

## Matrix multiplication

Each entry in the product is the dot product of a row of `A` and a column of `B`. For these two-dimensional arrays, `np.dot(A, B)` and `A @ B` give the same result.

```python
G = A @ B
G
```

## Transpose

Transposing swaps rows and columns.

```python
A_T = A.T
A_T
```

## Determinant

For this 2 × 2 matrix, the determinant is `1 * 4 - 2 * 3 = -2`.

```python
det = np.linalg.det(A)
det
```

## Inverse

Multiplying an invertible matrix by its inverse gives the identity matrix, up to floating-point rounding.

```python
A_inv = np.linalg.inv(A)
A_inv
```

## Trace

The trace is the sum of the diagonal entries.

```python
trace = np.trace(A)
trace
```

## Rank

The rank counts linearly independent rows or, equivalently, linearly independent columns.

```python
rank = np.linalg.matrix_rank(A)
rank
```

## Reductions

Use `axis=1` to sum across columns and get one total per row. Use `axis=0` to sum down the rows and get one total per column.

```python
total_sum = np.sum(A)


sum_rows = np.sum(A, axis=1)
sum_cols = np.sum(A, axis=0)


min_val = np.min(A)
max_val = np.max(A)


mean_val = np.mean(A)

print("Row totals:", sum_rows)
print("Column totals:", sum_cols)
```
