# Lesson 4: Matrix arithmetic homework

Compute the sum, difference, and matrix products for two sets of matrices. In each set, `A` and `B` are 3 × 3, and `C` is 3 × 2. The product `A @ B @ C` therefore has shape 3 × 2.

**Data:** No external dataset. Matrices are defined in the notebook.

## First set

```python
import numpy as np

matrix_A = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])

matrix_B = np.array([
    [-1, 0, 3],
    [2, -6, 1],
    [-7, 1, 5]
])

matrix_C = np.array([
    [-2, 3],
    [4, 0],
    [7, -1]
])

print(matrix_A + matrix_B)
```

```python
print(matrix_B - matrix_A)
```

```python
print(np.dot(matrix_A, matrix_B))
```

```python
print(matrix_A @ matrix_B @ matrix_C)
```

## Second set

Repeat the same four operations with a new set of matrices.

```python
matrix_A = np.array([
    [0, 3, 4],
    [-7, 1, -2],
    [2, -1, 3]
])

matrix_B = np.array([
    [1, 2, 4],
    [7, 0, -2],
    [-3, -2, 5]
])

matrix_C = np.array([
    [-3, 2],
    [0, 4],
    [-1, 7]
])

print("A + B =", matrix_A + matrix_B, sep="\n")
print("B - A =", matrix_B - matrix_A, sep="\n")
print("A @ B =", matrix_A @ matrix_B, sep="\n")
print("A @ B @ C =", matrix_A @ matrix_B @ matrix_C, sep="\n")
```
