# Multilingual speech synthesis with XTTS

Generate Russian speech from `text.txt` using the reference recording in `voice_reference.wav`. The sample is a 15-second excerpt from [a Russian poetry reading by Andrew Krizhanovsky](https://commons.wikimedia.org/wiki/File:Ru-Rodine_(Drozhzhin).ogg), released into the public domain by the speaker. The generated speech is written to `output.wav`.

Install a compatible Coqui TTS and PyTorch environment; the first run downloads model weights. The saved audio was generated on Kaggle with XTTS v2 and the reference recording above. The cells themselves were not executed in this notebook. Check the model’s [license](https://huggingface.co/coqui/XTTS-v2) before reuse.

**Data:** No dataset. The reference is an excerpt of [Andrew Krizhanovsky’s recording](https://commons.wikimedia.org/wiki/File:Ru-Rodine_(Drozhzhin).ogg); the example text is in `text.txt`.

```python
from TTS.api import TTS
import torch

device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"

tts = TTS(
    model_name="tts_models/multilingual/multi-dataset/xtts_v2",
    progress_bar=True
).to(device)
```

```python
with open("text.txt", "r", encoding="utf-8") as f:
    text = f.read()

print(text[:500])
```

```python
tts.tts_to_file(
    text=text,
    speaker_wav="voice_reference.wav",
    language="ru",
    file_path="output.wav"
)
```
